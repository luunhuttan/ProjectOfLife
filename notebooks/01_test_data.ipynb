```python
# Đường dẫn tương đối: lùi lại 1 cấp (../) ra khỏi thư mục notebooks, rồi chui vào thư mục data
record_path = '../data/mit-bih-arrhythmia-database-1.0.0/100'

# Dùng wfdb để đọc file dữ liệu (không cần ghi đuôi .dat hay .hea)
record = wfdb.rdrecord(record_path)

# In ra một vài thông tin cơ bản của bản ghi này
print("Tên bản ghi:", record.record_name)
print("Tần số lấy mẫu (fs):", record.fs, "Hz")
print("Số lượng kênh tín hiệu:", record.n_sig)
print("Tên các kênh:", record.sig_name)
```

```python
# Đường dẫn vẫn như cũ
record_path = '../data/mit-bih-arrhythmia-database-1.0.0/100'

# Lần này mình dùng wfdb.rdann để đọc file nhãn (đuôi .atr)
annotation = wfdb.rdann(record_path, 'atr')

# In ra thử 15 nhãn đầu tiên của bản ghi này
print("15 vị trí (mốc thời gian) đầu tiên bác sĩ đánh dấu:")
print(annotation.sample[:15])

print("\n15 ký hiệu (loại nhịp tim) tương ứng:")
print(annotation.symbol[:15])
```

```python
# Cắt 3000 điểm dữ liệu đầu tiên của bản ghi và file nhãn (khoảng 8.3 giây)
record_short = wfdb.rdrecord(record_path, sampto=3000)
annotation_short = wfdb.rdann(record_path, 'atr', sampto=3000)

# Yêu cầu thư viện wfdb vẽ đồ thị kết hợp cả sóng và nhãn
wfdb.plot_wfdb(record=record_short, 
               annotation=annotation_short, 
               time_units='seconds',
               figsize=(12, 6),
               title= 'sóng điện tim (8 giây đầu tiên - Bản ghi 100)')
```
